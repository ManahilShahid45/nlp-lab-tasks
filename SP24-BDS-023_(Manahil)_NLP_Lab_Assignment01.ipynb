## Task 1: Regex Fundamentals (Character Classes and Quantifiers)

For this task I am writing one sample text that has normal words, numbers, and a mix of capital and lowercase letters. Then I am writing four separate regex patterns on this same text.

Part (a) finds every digit sequence in the text. Part (b) finds every word that starts with a capital letter. Part (c) finds every sequence of exactly four digits, which is useful for picking out years. Part (d) finds every word that is between 3 and 6 characters long.

```python
text = "In 2023, I met Dr. Ahmed and Sara at a cafe. There were 45 apples, 12 boxes, and 3 pens on sale for $500. The event took place in 2024 near 1990s style buildings."

digits = re.findall(r'\d+', text)
print("Digits found:", digits)

capital_words = re.findall(r'\b[A-Z][a-z]*\b', text)
print("Capitalized words:", capital_words)

four_digit = re.findall(r'\b\d{4}\b', text)
print("Four digit sequences:", four_digit)

three_to_six = re.findall(r'\b[a-zA-Z]{3,6}\b', text)
print("Words 3 to 6 characters long:", three_to_six)
```

### What the output shows

The digits list shows every plain number in the text, including the year 2023, the quantities 45, 12 and 3, the price 500, the year 2024 and the number 1990. The regex \d+ simply means one or more digit characters in a row.

The capitalized words list shows every word that begins with an uppercase letter, such as In, Dr, Ahmed, Sara and The. The pattern [A-Z][a-z]* means one capital letter followed by zero or more lowercase letters.

The four digit sequences list only keeps 2023 and 2024, because these are the only two four digit numbers in the whole text. 1990 was not picked up here because it is followed directly by the letter s, so the word boundary rule \b does not match at exactly four digits in that spot.

The last list shows every word made only of letters that is between 3 and 6 characters long, such as met, Ahmed, cafe, apples, boxes and event. Shorter words like In and a, and longer words like buildings, are correctly left out.

## Task 2: Pattern Based Extraction (Real World Data)

Here I am writing a paragraph that mixes emails, phone numbers, order dates, and percentages together, the way real data often looks. Then I am writing one regex pattern for each of the four things I need to pull out, and I am using re.findall() for all of them as the assignment asks.

```python
paragraph = """Please contact us at support@example.com or sales@company.pk for help.
You can also call 0301-1234567 or 0333-9876543 during office hours.
Your order ID ORD12345 was placed on 15/08/2023 and confirmed on 20/09/2024.
The discount applied was 45% while the tax rate was 12.5%."""

emails = re.findall(r'[\w.+-]+@[\w-]+\.[\w.-]+', paragraph)
print("Emails found:", emails)

phones = re.findall(r'03\d{2}-\d{7}', paragraph)
print("Phone numbers found:", phones)

dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', paragraph)
print("Dates found:", dates)

percentages = re.findall(r'\b\d+(?:\.\d+)?%', paragraph)
print("Percentages found:", percentages)
```

### What the output shows

The emails list correctly pulls out both support@example.com and sales@company.pk. The pattern looks for normal characters before the @ sign, then normal characters after it, then a dot followed by more letters, which is the general shape of an email address.

The phone numbers list pulls out 0301-1234567 and 0333-9876543. The pattern 03 followed by two digits, a dash, and seven digits matches the exact format given in the assignment.

The dates list pulls out 15/08/2023 and 20/09/2024, since the pattern is looking for two digits, a slash, two digits, a slash, and four digits, which is the DD/MM/YYYY shape.

The percentages list pulls out 45% and 12.5%, showing that the pattern can handle both a plain whole number and a number with a decimal point before the percent sign.

## Task 3: Word Tokenization (NLTK vs spaCy)

For this task I am writing my own paragraph that has more than three sentences and includes contractions like don't and it's, as the assignment asks. I am also adding a hyphenated word and a dollar amount so that the difference between the two tokenizers becomes clearly visible.

Tokenization simply means breaking a paragraph down into smaller pieces, usually words and punctuation marks, so that a computer can process them one at a time.

```python
from nltk.tokenize import word_tokenize

paragraph = "I don't think it's going to rain today, but I'm not 100% sure. Sarah's plan is to visit New York-based markets, and she can't wait to buy things worth $50."

nltk_tokens = word_tokenize(paragraph)
print("NLTK tokens:")
print(nltk_tokens)
print("NLTK token count:", len(nltk_tokens))

nlp_blank = spacy.blank("en")
doc = nlp_blank(paragraph)
spacy_tokens = [token.text for token in doc]
print()
print("spaCy tokens:")
print(spacy_tokens)
print("spaCy token count:", len(spacy_tokens))

# Differences I noticed between the two outputs:
# 1. On the hyphenated word New York-based, NLTK keeps York-based as one single token, while spaCy splits it into three separate tokens: York, the hyphen, and based.
# 2. Because of this split, spaCy ends up producing two more tokens than NLTK for the exact same paragraph, even though both tools split contractions like don't and it's in a similar way.
```

### What the output shows

Both tokenizers split words like don't into do and n't, and it's into it and 's, so on contractions the two tools behave in a similar way here.

The real difference shows up on the word New York-based. NLTK treats York-based as one single token and keeps the hyphen inside it, while spaCy breaks it into three tokens, York, the hyphen by itself, and based. Because of this one difference, spaCy produces two more tokens overall than NLTK, even though the paragraph is the same for both.

This shows that even two well known tokenizers do not always agree on every word, especially around hyphens and other punctuation marks that are not simple word boundaries.

## Task 4: Token Attributes (Categorizing Tokens)

Here I am writing one sentence that has normal words, a number, an email address and a currency amount, exactly as the assignment describes. I am loading this sentence into a spaCy Doc object, which is spaCy's way of storing a piece of text along with information about every token inside it.

Then I am looping through every token and printing three attributes. is_alpha tells me whether a token is made only of letters. like_num tells me whether a token looks like a number. like_email tells me whether a token looks like an email address. Finally I am picking out only the tokens where is_currency is True, which marks currency symbols like the dollar sign.

```python
nlp = spacy.load("en_core_web_sm")

sentence = "I paid $50 and my friend sent an email to sara2024@gmail.com about 12 items."

doc = nlp(sentence)

print("Token and is_alpha:")
for token in doc:
    print(token.text, "->", token.is_alpha)

print()
print("Token, like_num and like_email:")
for token in doc:
    print(token.text, "-> like_num:", token.like_num, "| like_email:", token.like_email)

print()
currency_tokens = [token.text for token in doc if token.is_currency]
print("Tokens where is_currency is True:", currency_tokens)
```

### What the output shows

In the is_alpha list, ordinary words such as I, paid, and, my, friend and items are marked True, because they are made only of letters. The dollar sign, the number 50, the email address and the number 12 are all marked False, because none of them are made purely of letters.

In the like_num and like_email list, 50 and 12 are marked True for like_num, since they look like numbers, while everything else is False for that attribute. The email address sara2024@gmail.com is the only token marked True for like_email, since it is the only token that looks like an email address.

The final line shows that the dollar sign is the only token in this sentence where is_currency is True, which makes sense because it is the only currency symbol present.

## Task 5: Sentence Segmentation (Handling Tricky Cases)

For this last task I am writing a paragraph with more than one abbreviation in the middle of sentences, such as Dr., Mr. and e.g., which are exactly the tricky cases the assignment is asking about. Sentence segmentation means splitting a full paragraph into individual sentences.

I am loading the trained spaCy pipeline en_core_web_sm, which is a small pipeline that has actually learned from real text, unlike the blank pipeline I used in Task 3. Then I am using doc.sents to get the individual sentences, and I am printing how many sentences were found in total.

```python
nlp = spacy.load("en_core_web_sm")

paragraph = "Dr. Ahsan works at the city hospital. He usually arrives by 8 a.m. and leaves late. Mr. Khan, his colleague, joined the team last year. They both enjoy their work e.g. helping patients and teaching students."

doc = nlp(paragraph)

sentences = list(doc.sents)

for i, sent in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sent.text}")

print()
print("Total number of sentences detected:", len(sentences))
```

### What the output shows

The paragraph is correctly split into four sentences, even though it contains the abbreviations Dr., a.m., Mr. and e.g. right in the middle of sentences. None of these abbreviations caused an incorrect sentence break, which means the trained spaCy pipeline recognised the dot after each abbreviation as part of the abbreviation itself, and not as the end of a sentence.

This is different from what would happen with a very simple rule that just splits text on every single period, since that kind of simple rule would have wrongly cut Dr. Ahsan into two pieces at the dot after Dr. The trained pipeline avoids this mistake because it has learned common abbreviation patterns from real text during its training.
