## Task 04: Sets for Unique Vocabulary

### (a) Removing duplicates with a set

Converting a list to a set removes any repeated values because a set only stores unique elements.

```python
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]
vocab = set(raw)

print(raw)
print(vocab)
print(len(vocab))

# Sets remove duplicates automatically because internally a set only keeps one copy of each distinct value. When a duplicate item is added, Python checks if it already exists and simply ignores the new copy.
```

### (b) Union, intersection, and difference

Sets support mathematical operations such as union, intersection, and difference, which are useful for comparing vocabularies.

```python
set_A = {"machine", "learning", "data", "model"}
set_B = {"deep", "learning", "neural", "data"}

union_set = set_A | set_B
intersection_set = set_A & set_B
difference_set = set_A - set_B

print(union_set)
print(intersection_set)
print(difference_set)
```

### (c) Removing stopwords with set difference

Stopwords are common words that carry little meaning. Using set difference lets me remove them from a token list in one step.

```python
stopwords = {"is", "the", "a", "in", "of", "and", "to"}
tokens = ["nlp", "is", "a", "field", "of", "ai", "and", "the", "future"]

cleaned_tokens = set(tokens) - stopwords
print(cleaned_tokens)
```

### (d) add and discard methods

The `add()` method inserts a new element and `discard()` removes an element without raising an error if it is missing.

```python
vocab.add("transformer")
vocab.add("bert")
vocab.discard("python")

print("ai" in vocab)
print(vocab)
```

## Task 06: String Cleaning Methods

### (a) strip, lower, replace

I applied three cleaning steps one after another and printed the result at each stage.

```python
text = "   MACHINE Learning is AMAZING!!!   "

stripped = text.strip()
print(stripped)

lowered = stripped.lower()
print(lowered)

no_exclaim = lowered.replace("!", "")
print(no_exclaim)
```

### (b) split, list comprehension, join

I tokenised the cleaned text, uppercase each token using a list comprehension, and joined the tokens back into a single string.

```python
words = no_exclaim.split()
print(words)

upper_words = [w.upper() for w in words]
print(upper_words)

joined = " | ".join(upper_words)
print(joined)
```

### (c) replace and find

### (d) A reusable normalize function

I combined several cleaning steps into a single reusable function and tested it on more than one messy string.

```python
def normalize(text):
    cleaned = text.strip().lower()
    cleaned = cleaned.replace(",", "").replace(".", "")
    return cleaned

print(normalize("  Hello, World.  "))
print(normalize("NLP is Fun, and Useful."))
```

### (b) Classifying tokens

I looped through a list of tokens and label each one as alphabetic, numeric, or mixed.

```python
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        print(f"{token}: alphabetic")
    elif token.isdigit():
        print(f"{token}: numeric")
    else:
        print(f"{token}: mixed")
```

### (c) Grouping tokens

I separated a list of tokens into three groups based on whether they are pure words, pure numbers, or mixed.

```python
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print(pure_words)
print(pure_numbers)
print(mixed_tokens)
```

### (d) A reusable validate_token function

I wrapped the classification logic into a function and applied it to a new list of tokens.

```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

sample_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for token in sample_tokens:
    print(token, "->", validate_token(token))
```

## Task 08: Functions and Lambda for NLP Pipelines

### (a) A function and an equivalent lambda

I wrote a normal function and a lambda that both strip and lowercase a string, then compared their output.

```python
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample = "  HELLO Natural Language Processing  "
print(clean(sample))
print(clean_lambda(sample))
```

### (b) Tokenising and filtering short words

I wrote one function to tokenise text and another to remove tokens shorter than a given length, then chained them together.

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence3 = "I love working with NLP tools in Python daily"
result = remove_short(tokenize(sentence3), 3)
print(result)
```

### (c) Building a word length dictionary

I wrote a function that maps each token to its length and returns the result as a dictionary.

```python
def word_lengths(tokens):
    return {t: len(t) for t in tokens}

sample_tokens = ["language", "model", "ai", "processing", "deep"]
print(word_lengths(sample_tokens))
```

### (d) Combining everything into one pipeline

I combined cleaning, tokenising, filtering, and length mapping into a single pipeline function and tested it on two sentences.

```python
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

print(nlp_pipeline("Natural Language Processing is an exciting field", 3))
print(nlp_pipeline("Deep learning models require a lot of data", 2))
```

## Task 09: map, filter, sorted and Counter

### (a) Using map with a lambda

I used `map()` to lowercase every word and separately to compute the length of every word.

```python
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lower_words = list(map(lambda w: w.lower(), words))
word_lens = list(map(lambda w: len(w), words))

print(lower_words)
print(word_lens)
```

### (b) Using filter with a lambda

I filtered the word list to keep only long words, and separately to keep only words containing the letter t.

```python
words_long = list(filter(lambda w: len(w) > 3, words))
words_with_t = list(filter(lambda w: "t" in w.lower(), words))

print(words_long)
print(words_with_t)
```

### (c) Sorting in three different ways

I sort the same list alphabetically, by length ascending, and by length descending.

```python
alphabetical = sorted(words)
by_length_asc = sorted(words, key=len)
by_length_desc = sorted(words, key=len, reverse=True)

print("Alphabetical:", alphabetical)
print("By length ascending:", by_length_asc)
print("By length descending:", by_length_desc)
```

### (d) Counting word frequency with Counter

The `Counter` class makes it simple to count how often each word appears in a piece of text.

```python
from collections import Counter

sentence4 = "nlp nlp python ai nlp python text ai nlp python text"
words4 = sentence4.split()
freq = Counter(words4)

print(freq)
print(freq.most_common(3))
print(freq["python"])
print("Total words:", len(words4))
print("Unique vocabulary size:", len(freq))
```
